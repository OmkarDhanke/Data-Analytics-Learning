## Attribute: The Value of an Object 
### A Field in a record , defined by its class

```python
class JobPost:
    def __init__(self, job_title, location, salary):
        self.job_title = job_title
        self.location = location
        self.salary = salary
```

```python
job_1 = JobPost(job_title, job_location, job_salary)
```

```python
job_1.job_title()
```

## Methods: The Function of an Object
### A function defined inside a class that oprrates on its object

```python
class JobPost:
    def __init__(self, job_title, location, salary):
        self.job_title = job_title
        self.location = location
        self.salary = salary

    def Display_Job_Details(self):
        print(f"JOB:           {self.job_title} \nlocated in:    {self.location} \nsalary:        {self.salary}")
```

```python
job_1 = JobPost(job_title, job_location, job_salary)
job_1.job_title
```
