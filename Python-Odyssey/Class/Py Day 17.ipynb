```python
# Write a method in side a class feach volum of spear by given
import math

class Sphere:
    def area(self, r):
        self.radius = r

    def calculate(self):
        self.volume = (4/3) * math.pi * self.radius ** 3

    def display(self):
        return f"The radius is {self.radius} and the volume of the sphere is {self.volume:.2f}"

z = Sphere()
z.area(2)
z.calculate()
print(z.display())
```
