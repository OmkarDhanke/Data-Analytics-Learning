# 1. Class


```python
# In Class When we create a variable in class it become atribute
class Human():
  hair = "Black"
  nose = "Sharp"
  eyes = "Brown"
  height = 5.8
  weight = 59

print(Human.hair)
print(Human.nose)
print(Human.eyes)
print(Human.height)
print(Human.weight)
```

# 2. Methods


```python
# Methods
class Human():
  hair = "Black"
  nose = "Sharp"
  eyes = "Brown"
  height = 5.8
  weight = 59

  def speak(self):
    return "I can Speak"

  def walk(self):
    return "I can walk"

print(Human.eyes)
print(Human.speak("a"))
```

```python
# Creat a method in side a class and add One thausand Bonus for given salary
class Myself():
  department = "Sales"

  def detail(self,name,salary):
    self.name = name
    self.salary = salary

  def calculation(self):
    return self.salary + 1000

  def display(self):
    print(f"name : {self.name}\n"
          f"department : {self.department}\n"
          f"salary : {self.salary}\n"
          f"bonus : {self.calculation()}")

s = Myself()
s.detail("Omkar",50000)
s.display()
```
