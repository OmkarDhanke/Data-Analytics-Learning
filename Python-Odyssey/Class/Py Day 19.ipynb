# **Inheritence**
### Inheritance in Python allows one class (child) to use the properties and methods of another class (parent).It promotes code reusability, organization, and helps create a clear relationship between classes.

```python
class Mom():
  Eyes = 'Blue'
  Voice = 'Sweet'

  def Intro(self):
    return "I am mom Class"

class Dad():
  Nose = 'Sharp'
  Eyes = 'Brown'
  height = 5.9
  salary = 25000

  def Intro(self):
    return "I am dad Class"

class Son(Mom,Dad): # These are the classes asise to the son class
  def Intro(self):
    return "I am son Class"

x = Son()
print(x.Eyes) # when we call the son eyes it check the fist calss that check the atribute and if it is there it return the value this is calledx method overloading

print(x.salary)
print(x.Intro())
```

### What is Polymorfisam?

It allows one interface to be used for different data types or classes.
In Python, it’s often seen through method overriding and operator overloading.

### Method Overloading

Definition:
In Python, method overloading means using one method name for different tasks by changing the number or type of arguments. Python doesn’t support true overloading, but it can be done using default arguments or *args.

Types:
1. Compile-time Overloading: Happens in other languages like Java/C++, decided  before running the code.
2. Pseudo Overloading: In Python, done using default or variable arguments.

```python
# method overloading
class Mom():
  Eyes = 'Blue'
  Voice = 'Sweet'

  def Intro(self):
    return "I am mom Class"

class Dad():
  Nose = 'Sharp'
  Eyes = 'Brown'
  height = 5.9
  salary = 25000

  def Intro(self):
    return "I am dad Class"

class Son(Mom,Dad): # These are the classes asise to the son class
  def Intro(self):
    return "I am son Class"

x = Son()
print(x.Eyes)
print(x.salary)
print(x.Intro())
```

### Method Overriding

Definition:
When a child class has a method with the same name as its parent class, it overrides or changes the parent’s version.

Types:
1. Simple Overriding: The child replaces the parent’s method completely.
2. Dynamic Overriding: The child adds extra work but still uses the parent’s method with super().

```python
# Method Overriding Example

# Parent Class 1
class Mom():
    Eyes = 'Blue'
    Voice = 'Sweet'

    def Intro(self):  # Method in Mom class
        return "I am mom Class"


# Parent Class 2
class Dad():
    Nose = 'Sharp'
    Eyes = 'Brown'      # Same attribute name as in Mom (will show inheritance order effect)
    height = 5.9
    salary = 25000

    def Intro(self):    # Method in Dad class
        return "I am dad Class"


# Child Class inheriting from both Mom and Dad
class Son(Mom, Dad):  # Order matters — Mom first, Dad second
    Eyes = 'Blank'     # Overriding the Eyes attribute from parents

    def Intro(self):   # Overriding the Intro() method from parents
        return "I am son Class"


# Object creation
x = Son()              # Creates an instance of the Son class

# Accessing attributes and methods
print(x.Eyes)          # Looks for Eyes in Son → found → prints 'Blank'
print(x.salary)        # Not found in Son or Mom → found in Dad → prints 25000
print(x.Intro())       # Intro() overridden in Son → prints "I am son Class"
```

## 1. Single Inheritance

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):  # Employee inherits from Person
    def show_role(self):
        print(self.name, "is an employee")

emp = Employee("Sarah")
print("Name:", emp.name)
emp.show_role()
```

# 2. Multiple Inheritance

```python
class Person:
    def __init__(self, name):
        self.name = name

class Job:
    def __init__(self, salary):
        self.salary = salary

class Employee(Person, Job):  # Inherits from both Person and Job
    def __init__(self, name, salary):
        Person.__init__(self, name)
        Job.__init__(self, salary)

    def details(self):
        print(self.name, "earns", self.salary)

emp = Employee("Jennifer", 50000)
emp.details()
```

# 3. Multilevel Inheritance

```python
class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def show_role(self):
        print(self.name, "is an employee")

class Manager(Employee):  # Manager inherits from Employee
    def department(self, dept):
        print(self.name, "manages", dept, "department")

mgr = Manager("Joy")
mgr.show_role()
mgr.department("HR")
```
