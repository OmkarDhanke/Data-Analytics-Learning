```python
area = [600,800,1000,1200,1500]
price = [15,18,25,35,40]
```

```python
# Title Function
plt.plot(area,price)
plt.title("Area vs Price")
```

```python
# Lable
plt.plot(area,price)
plt.title("Area vs Price")
plt.xlabel("Area")
plt.ylabel("Price")
```

```python
# Grid
plt.plot(area,price)
plt.title("Area vs Price")
plt.xlabel("Area")
plt.ylabel("Price")
plt.grid(True)
```

```python
# xlim and ylim to define x and y range
# What are the useage of lim
plt.plot(area,price)
plt.title("Area vs Price")
plt.xlabel("Area")
plt.ylabel("Price")

plt.grid(True)

plt.xlim(400,2000)
plt.ylim(10,50)
```

```python
# xticks and yticks to rotate lables
# what dose this function do
plt.plot(area,price)

plt.title("Area vs Price")
plt.xlabel("Area")
plt.ylabel("Price")

plt.xticks(rotation = 45)
plt.yticks(rotation = 90)

plt.grid(True)
```

# Multi Plot

```python
x = np.arange(1,20)
y1 = np.sqrt(x)
y2 = np.log(x)

plt.plot(x,y1, color = "blue")
plt.plot(x,y2, color = "red")
plt.show()
```

```python
x = np.arange(1,20)
y1 = np.sqrt(x)
y2 = np.log(x)

plt.plot(x,y1, color = "blue")
plt.plot(x,y2, color = "green")
plt.legend(labels = ["Square root","log"])
plt.grid(True)
```

```python
x = [1,2,3]
y = [5,7,4]

x1 = [1,2,3]
y1 = [11,14,22]

plt.plot(x,y, label='First Line',color = "red")
plt.plot(x1,y1, label = "second Line",color = "blue")

plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```

# Bar chat & Histogram

```python
x , y = [2,4,6,8,10],[6,7,8,2,4]
plt.bar(x,y, label = "bar1", color = "teal")
plt.show()
```

```python
x , y = [2,4,6,8,10],[6,7,8,2,4]
plt.bar(x,y, label = "bar1", color = "c")
plt.bar(x,y, label = "bar1", color = "r")

plt.title("Bar Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
x,y = [2,4,6,8,10],[6,7,8,2,4]
x1,y1 = [1,2,3,4,5], [7,8,2,4,2]

plt.bar(x,y, label = "bar1", color= "teal")
plt.bar(x1,y1, label = "bar2", color= "blue")

plt.title("Bar Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```
