  ## Filter
   

```python
a = [1,'a',' ',0,None,' ',3,0,None]
filter(None,a)
```

```python
a = [1,'a','',0,None,2,'',3,0,None]
list(filter(None,a))
```

```python
a = [1,'a','',0,None,2,'',3,0,None]
b = list(filter(None,a))
b = [1, 'a', 2, 3]
c = list(filter('a',b))

print(c)
```

```python
x = [1,'a',2,3,(1,None,2)]
list(filter(None,x))
```

## Map

```python
a = [1,2,3,5.5,26]

def vos(r):
  x = 3/4 * 3.14 * (r**3)
  return x

map(vos,a)
c = list(map(vos,a))
print(c)
```
