## Constructor



```python
a = 'world' # Global attribute

class ABC():
  a = 'india' # Local attribute
  def __init__(self,name,salary,age):

    self.name = name
    self.salary = salary # instance
    self.age = age

print(a)
print(ABC.a)
```

## Abstraction

```python
class Abst():
  a = 10
  _b = 20
  __c = 30

x = Abst()
print(x.a) # Public
print(x._b) # Protectet
print(x.__c) # Privet
```

```python
class Maramari():
  def __init__(self,name):
    self.name = name
    self.__life = 3
    self.__score = 0

m = Maramari('Maramari')
print(m.name)
print(m._Maramari__life)
print(m._Maramari__score)
```

```python
class WWE():
  def __init__(self,name):
    self.name = name
    self.__life = 3
    self.__score = 0

  def Punch(self):
    self.__score = self.__score + 10
    return self.__score

  def Kick(self):
    self.__score = self.__score + 15
    return self.__score

  def Slap(self):
    self.__score = self.__score + 7
    return self.__score

  def hand_gaurd(self):
    self.__score = self.__score - 2
    return self.__score

  def Dodge(self):
    self.__life = self.__life + 1
    return self.__life

  def Stabbed(self):
    self.__life = self.__life - 1
    return self.__life

  def Display_satus(self):
    print(f"Name:-{self.name}\n"
          f"Score:-{self.__score}\n"
          f"life:-{self.__life}")

player = WWE("John Cena")

# Perform some actions
print("Punch Score:", player.Punch())
print("Kick Score:", player.Kick())
print("Slap Score:", player.Slap())
print("Guard Score:", player.hand_gaurd())
print("Dodged Life:", player.Dodge())
print("Life after being stabbed:", player.Stabbed())
```
